print es una funcion de python que muestra un resultado en pantalla.

```python
edad = 25
nombre = "Francy"
altura = 1.72
is_estudiante = True

print(nombre)
print(edad)
print(altura)
print(is_estudiante)

print(type(nombre))
print(type(edad))
print(type(altura))
print(type(is_estudiante))
```

declaramos variables,una variable en python es una etiqueta que guarda un valor en memoria, este valor puede ser numeros enteros (int) o flotantes con decimales (float), cadenas de textos (str) y valores booleanos (True o False).
al declararla debemos tener en cuenta que los nombres sean descriptivos, letras minusculas, si tiene varias palabras estas deben estar separadas con guion bajo (_).
el operador igual (=) debe contar con un espacio en ambos lados (variable = valor).

```python
print("25")
print(25)

print(type("25"))
print(type(25))
```

Podemos observar que "25" es diferente de 25, el primero es una cadena de texto y el segundo un entero.
